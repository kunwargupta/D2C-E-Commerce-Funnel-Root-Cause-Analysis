# Phase 2: Funnel Analysis

```python
# 1. Count visitors at each stage

# Stage1: count of website visitors
visited = df['visited_website'].sum()

# Stage 2: count of product viewed
viewed = df['viewed_product'].sum()

# Stage 3: count of product added to cart
cart = df['added_to_cart'].sum()

# Stage 4: count of checkout
checkout = df['checkout_started'].sum()

# Stage 5: count of purchased
purchased = df['purchase_completed'].sum()
```

```python
# Calculate drop rates

drop_visited_to_viewed = ((visited - viewed) / visited) * 100
drop_viewed_to_cart = ((viewed - cart) / viewed) * 100
drop_cart_to_checkout = ((cart - checkout) / cart) * 100
drop_checkout_to_purchase = ((checkout - purchased) / checkout) * 100
```

```python
# 3.  Calculate retention %

viewed_retention = (viewed / visited) * 100
cart_retention = (cart / viewed) * 100
checkout_retention = (checkout / cart) * 100
purchase_retention = (purchased / checkout) * 100
```

```python
round(viewed_retention, 2), round(cart_retention, 2), round(checkout_retention, 2), round(purchase_retention, 2)
```

```python
# 4. Create results table

result = {
    "Stage": ["Visited", "Viewed", "Cart", "Checkout", "Purchased"],
    "Count": [visited, viewed, cart, checkout, purchased],
    "Drop": [np.nan, drop_visited_to_viewed, drop_viewed_to_cart, drop_cart_to_checkout, drop_checkout_to_purchase],
    "Retention": [np.nan, viewed_retention, cart_retention, checkout_retention, purchase_retention]
}
```

```python
funnel_df = pd.DataFrame(result)
funnel_df
```

```python

stages = {
    "visited": "visited_website",
    "viewed": "viewed_product",
    "cart": "added_to_cart",
    "checkout": "checkout_started",
    "purchased": "purchase_completed"
}

def stage_retention(data, group_cols):
    result = data.groupby(group_cols, as_index=False).agg({
        stages["visited"]: "sum",
        stages["viewed"]: "sum",
        stages["cart"]: "sum",
        stages["checkout"]: "sum",
        stages["purchased"]: "sum"
    })

    result = result.rename(columns={
        stages["visited"]: "visited",
        stages["viewed"]: "viewed",
        stages["cart"]: "cart",
        stages["checkout"]: "checkout",
        stages["purchased"]: "purchased"
    })

    result["visit_to_view_retention"] = (
        result["viewed"] / result["visited"] * 100
    )

    result["view_to_cart_retention"] = (
        result["cart"] / result["viewed"] * 100
    )

    result["cart_to_checkout_retention"] = (
        result["checkout"] / result["cart"] * 100
    )

    result["checkout_to_purchase_retention"] = (
        result["purchased"] / result["checkout"] * 100
    )

    result["conversion_rate"] = (
        result["purchased"] / result["visited"] * 100
    )

    percentage_cols = [
        "visit_to_view_retention",
        "view_to_cart_retention",
        "cart_to_checkout_retention",
        "checkout_to_purchase_retention",
        "conversion_rate"
    ]

    result[percentage_cols] = result[percentage_cols].round(2)

    return result
```

```python
channel_retention = stage_retention(df, ["channel"])
device_retention = stage_retention(df, ["device"])
user_type_retention = stage_retention(df, ["user_type"])
campaign_retention = stage_retention(df, ["campaign_type"])
region_retention = stage_retention(df, ["region"])
```

# Phase 4

#### Finding 1: View → Cart is the main bottleneck

-- Evidence: 65.1% drop

-- Interpretation: Product-page engagement does not translate into cart intent.

-- Root-cause hypothesis: weak product value communication, pricing friction, or poor product-page experience.

-- Recommendation: test product-page messaging, price visibility, reviews, CTA placement, and product recommendations.

-- Validation needed: compare view→cart by channel, campaign, device, and product/category if available.

#### Finding 2: Paid Ads underperform Email

-- Evidence: Paid Ads 6.72% vs Email 7.31%.

-- Interpretation: Paid traffic is less qualified or has weaker message-to-landing-page alignment.

-- Root-cause hypothesis: targeting, creative, or landing-page mismatch.

-- Recommendation: audit paid campaigns by audience, creative, landing page, and acquisition cost.

-- Do not claim paid ads caused the gap without campaign-level evidence.

#### Finding 3: Mobile has a smaller but consistent disadvantage

-- Evidence: Mobile 6.78% vs Desktop 6.91%.

-- Stronger evidence: Email + Desktop 8.13%; Paid Ads + Mobile 6.71%.

-- Recommendation: prioritize mobile checkout and product-page QA, especially for Paid Ads.

-- Avoid calling this a major bottleneck. The gap is small.

#### Finding 4: Returning users underperform New users

-- Evidence: Returning 6.62% vs New 6.92%; Paid Ads + Returning = 6.41%.

-- Root-cause hypothesis: weak reactivation offers, irrelevant targeting, or poor returning-user experience.

-- Recommendation: test personalized offers, remarketing exclusions, and returning-user landing pages.

-- Validate with repeat-session, offer, and time-since-last-purchase data.

# RCA statement:
